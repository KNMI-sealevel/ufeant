```python
import os
import matplotlib as mpl
import matplotlib.pyplot as plt
import xarray as xr
import numpy as np
```

```python
SMBtot = 2576
AMBtot = 300
SMB_gr = 2083
AMB_gr = 190

runs = ['spinup_hiv_u_notm_2','spinup_hiv_u_tm','spinup_hiv_u_tm_2']
#runs = ['ant_test_PIG_THW_CD_2','spinup_hiv_u_notm']
vars = ['gl_flux','BMB_gr','BMB_fl','LMB_total','ice_volume']
refvals = [-SMB_gr-AMB_gr,0,-.55*(SMBtot+AMBtot),-.45*(SMBtot+AMBtot),64.13]
```

```python
fig = plt.figure(figsize=(10,7))

gs = fig.add_gridspec(len(vars),1, height_ratios=np.ones(len(vars)),left=0.1, right=0.99, bottom=0.15, top=0.95, wspace=0.1, hspace=0.1)
Ax = {}
for v,var in enumerate(vars):
    Ax[v] = fig.add_subplot(gs[v])
    Ax[v].set_ylabel(var)
    Ax[v].axhline(refvals[v],0,1,c='k',ls=':')
    if v<len(vars)-1:
        Ax[v].set_xticklabels([])

for r,run in enumerate(runs):
    ds = xr.open_dataset(f'../output/{run}/scalar_output_ANT_00001.nc')
    for v,var in enumerate(vars):
        Ax[v].plot(ds.time,ds[var],label=run)
        Ax[v].scatter(ds.time[-1],ds[var][-1],zorder=10)

Ax[0].legend(ncol=3)

plt.savefig('../figures/timeseries.png',dpi=600)
```
